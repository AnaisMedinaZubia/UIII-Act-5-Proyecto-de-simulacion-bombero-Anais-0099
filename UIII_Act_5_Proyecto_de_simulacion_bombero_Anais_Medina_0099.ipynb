```python
# Simulación de IA para Asistente de Bomberos
# Este programa evalúa el nivel de riesgo y el método de extinción.

print("--- SISTEMA DE IA PARA EVALUACIÓN DE INCENDIOS ---")

# Entrada interactiva de datos (operadores de entrada) [2, 4]
temperatura = float(input("Ingrese la temperatura detectada en grados Celsius: "))
presencia_personas = input("¿Hay presencia de personas en el edificio? (si/no): ").lower()
tipo_combustible = int(input("Tipo de material (1: Madera/Papel, 2: Eléctrico, 3: Químico): "))

# 1. Evaluación de Prioridad (Operador relacional y lógico) [5, 6]
# Se determina si la situación es crítica combinando temperatura y vidas en riesgo.
if temperatura > 100 and presencia_personas == "si":
    prioridad = "CRÍTICA: Rescate inmediato requerido"
elif temperatura > 100 or presencia_personas == "si":
    prioridad = "ALTA: Despliegue de equipo de protección"
else:
    prioridad = "ESTÁNDAR: Asegurar el perímetro"

# 2. Decisión de Extinción (Estructura if elif else múltiple) [1, 2]
# La IA selecciona el agente extintor basándose en el tipo de combustible.
if tipo_combustible == 1:
    agente = "Agua a presión o polvos químicos clase A"
elif tipo_combustible == 2:
    # Para incendios eléctricos el agua es peligrosa (uso de lógica de seguridad)
    agente = "Extintor de CO2 o Polvo Químico Seco (No usar agua)"
elif tipo_combustible == 3:
    agente = "Espuma química o arena seca"
else:
    agente = "Desconocido. Realizar inspección visual antes de actuar"

# 3. Cálculo de Nivel de Peligro (Operadores aritméticos) [7, 8]
# Se asume un riesgo base y se incrementa según la temperatura.
riesgo_base = 50
nivel_peligro = riesgo_base + (temperatura * 0.5)

# Salida de resultados con diagnóstico de la IA [9]
print("\n--- DIAGNÓSTICO FINAL DE LA IA ---")
print(f"Estado de Prioridad: {prioridad}")
print(f"Agente recomendado: {agente}")

# Evaluación final del nivel de peligro (if-else simple) [5]
if nivel_peligro >= 150:
    print("ALERTA: El nivel de peligro supera el umbral seguro. Evacuar bomberos si es necesario.")
else:
    print(f"Nivel de riesgo calculado: {nivel_peligro:.2f} puntos. Proceder con cautela.")
    print("Programa realizado por Anais Medina NL = 40")
```
